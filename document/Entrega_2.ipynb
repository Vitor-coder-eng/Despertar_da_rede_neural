##**Aprofundando a personalização: Treinamento de uma CNN do zero**

Após os resultados limitados obtidos com a YOLO tradicional, percebemos a necessidade de adotar uma abordagem mais personalizada. Decidimos, então, treinar uma CNN (Rede Neural Convolucional) do zero, voltada especificamente para o nosso contexto.
A proposta é simples: construir um modelo do zero, mas com uma modificação importante no dataset utilizado nos modelos anteriores — agora com o objetivo de classificar os objetos em questão: corn e fox.
Para essa nova abordagem, foi necessário reorganizar o conjunto de dados, separando as imagens de treinamento, validação e teste conforme as labels.

###**Importando as bibliotecas necessárias.**

```python
import tensorflow as tf
from tensorflow.keras.preprocessing.image import ImageDataGenerator
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Conv2D, MaxPooling2D, Flatten, Dense, Dropout
import matplotlib.pyplot as plt
import os
```

###**Definindo os caminhos das pastas.**

```python
base_dir = '/content/drive/MyDrive/CNN_dataset'
train_dir = os.path.join(base_dir, 'train')
val_dir = os.path.join(base_dir, 'val')
test_dir = os.path.join(base_dir, 'test')
```

###**Preparando os dados para o modelo.**

```python
img_height, img_width = 150, 150
batch_size = 32

# Gerador para treino (com augmentations)
train_datagen = ImageDataGenerator(rescale=1./255, # Normalizando os valores dos pixels dividindo por 255
                                   rotation_range=20,
                                   zoom_range=0.2,
                                   horizontal_flip=True)

# Validação e teste apenas com normalização
val_datagen = ImageDataGenerator(rescale=1./255)
test_datagen = ImageDataGenerator(rescale=1./255)

# Gerar os dados
train_generator = train_datagen.flow_from_directory(
    train_dir,
    target_size=(img_height, img_width),
    batch_size=batch_size,
    class_mode='binary'  # porque temos 2 classes: corn e fox
)

val_generator = val_datagen.flow_from_directory(
    val_dir,
    target_size=(img_height, img_width),
    batch_size=batch_size,
    class_mode='binary'
)

test_generator = test_datagen.flow_from_directory(
    test_dir,
    target_size=(img_height, img_width),
    batch_size=batch_size,
    class_mode='binary',
    shuffle=False  # importante para avaliação posterior
)
```

A saída acima exibe a quantidade de imagens presentes em nosso dataset. Temos 64 imagens destinadas ao treinamento, 8 para validação e 7 para teste, todas distribuídas em duas classes: corn (milho) e fox (raposa).

###**Construindo a Rede Neural Convolucional (CNN).**

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Conv2D, MaxPooling2D, Flatten, Dense, Dropout

model = Sequential([
    Conv2D(32, (3, 3), activation='relu', input_shape=(img_height, img_width, 3)), # Aplicando filtros para extrair características como bordas e curvas
    MaxPooling2D(pool_size=(2, 2)), # Reduzindo a dimensão das imagens, mantendo os dados mais importantes

    Conv2D(64, (3, 3), activation='relu'),
    MaxPooling2D(pool_size=(2, 2)),

    Flatten(), # Achatando os dados para vetor 1D, preparando para as camadas densas
    Dense(64, activation='relu'), # Conectando totalmente as camadas, onde o modelo consegue aprender combinações complexas das características
    Dropout(0.5),
    Dense(1, activation='sigmoid')  # saída binária
])
```

###**Compilando o modelo.**

```python
model.compile(optimizer='adam',
              loss='binary_crossentropy',
              metrics=['accuracy'])
```

###**Treinando o modelo.**

```python
history = model.fit(
    train_generator,
    epochs=10,
    validation_data=val_generator
)
```

###**Explicação do código:**
Para este modelo, configuramos o treinamento para 10 épocas. Como se trata de um dataset pequeno, essa quantidade é suficiente para que o modelo aprenda sem correr risco de overfitting, ou seja, memorizar os dados de treino em vez de generalizar bem para novos dados.

###**Avaliando o modelo (acurácia e perda)**

```python
import matplotlib.pyplot as plt

# Acurácia
plt.plot(history.history['accuracy'], label='Treino')
plt.plot(history.history['val_accuracy'], label='Validação')
plt.title('Acurácia do modelo')
plt.xlabel('Épocas')
plt.ylabel('Acurácia')
plt.legend()
plt.show()

# Perda
plt.plot(history.history['loss'], label='Treino')
plt.plot(history.history['val_loss'], label='Validação')
plt.title('Perda do modelo')
plt.xlabel('Épocas')
plt.ylabel('Loss')
plt.legend()
plt.show()
```

**Análise dos Gráficos de Acurácia e Perda**

Ao analisarmos os gráficos gerados durante o treinamento da CNN, observamos comportamentos interessantes que nos ajudam a entender melhor o desempenho do modelo.

Acurácia:
Como esperado, a acurácia teve um crescimento nas primeiras épocas, alcançando um pico por volta da época 4. Em seguida, houve uma queda acentuada na validação, seguida de uma nova subida, formando um padrão semelhante à letra **"M"** no gráfico. Esse comportamento pode estar relacionado à instabilidade no aprendizado, causada provavelmente pelo tamanho reduzido do dataset, o que pode levar o modelo a se ajustar demais (overfitting) a certos dados. Ainda assim, o modelo voltou a alcançar 100% de acurácia em épocas posteriores, o que pode indicar que ele conseguiu capturar bem os padrões, apesar das oscilações.

Perda:
No gráfico de perda, notamos um comportamento oposto: inicialmente, a perda teve um leve aumento — o que é comum no início do treinamento, enquanto o modelo busca ajustar seus pesos — e, em seguida, apresentou uma queda consistente, oscilando ao longo das épocas. Essa redução gradual indica que o modelo foi conseguindo minimizar o erro nas suas predições.

Conclusão:
O comportamento observado no gráfico **"Perda do modelo"**, principalmente no treino, remete ao conceito de **“ponto de cotovelo”**, onde o modelo deixa de apresentar melhorias significativas com o avanço das épocas. Essa análise reforça a importância de ajustar o número de épocas de acordo com o desempenho observado, evitando tanto o underfitting quanto o overfitting.

###**Mostrando as imagens com a predição**.

```python
import matplotlib.pyplot as plt
import numpy as np
from tensorflow.keras.preprocessing import image

def predict_image(img_path):
    img = image.load_img(img_path, target_size=(img_height, img_width))
    img_array = image.img_to_array(img)
    img_array = np.expand_dims(img_array, axis=0) / 255.0

    prediction = model.predict(img_array)[0][0]
    class_name = "Raposa" if prediction >= 0.5 else "Milho"

    # Mostrar a imagem com a predição
    img_display = image.load_img(img_path)
    plt.imshow(img_display)
    plt.title(f"Predição: {class_name} ({prediction:.2f})")
    plt.axis("off")
    plt.show()
```

```python
predict_image('/content/drive/MyDrive/CNN_dataset/test/corn/photo 37.png')
predict_image('/content/drive/MyDrive/CNN_dataset/test/fox/photo 38_1.png')
predict_image('/content/drive/MyDrive/CNN_dataset/test/corn/photo 38.png')
predict_image('/content/drive/MyDrive/CNN_dataset/test/fox/photo 37_1.png')
predict_image('/content/drive/MyDrive/CNN_dataset/test/fox/photo 39_1.png')
```

###**Conclusão e Análise Crítica**

Durante os testes, foi possível observar que o modelo teve **maior precisão na classificação das imagens de raposa**. Em alguns exemplos, imagens de raposa foram corretamente identificadas com **alta confiança**, como nos casos com 0.97 e 0.77 de probabilidade para a classe correta. Já as imagens de milho, embora também tenham sido classificadas corretamente, apresentaram **valores de confiança bem menores**, como 0.03 e 0.08. Esse comportamento indica que a rede teve **mais facilidade em aprender os padrões visuais da classe "fox"** do que da classe **"corn"**.

Um experimento curioso foi realizado com uma imagem propositalmente montada contendo **uma raposa com um milho na boca**. O modelo classificou essa imagem como **milho com 0.21 de confiança**, mostrando certa sensibilidade ao objeto "milho", mas confundindo o contexto geral — o que revela uma **limitação natural para situações ambíguas ou múltiplas classes em uma imagem só**.
