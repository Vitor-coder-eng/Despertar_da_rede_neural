### Testando o modelo. Nesta etapa usamos um conjunto de imagens inéditas para avaliarmos seu desempenho e como se comporta em ambientes menos controlados.

```python
import os
import subprocess

def get_latest_train_run_folder():
    subfolders = [f.path for f in os.scandir('yolov5/runs/train') if f.is_dir()]
    latest_folder = max(subfolders, key=os.path.getctime, default=None)
    return latest_folder

latest_run = get_latest_train_run_folder()
result = subprocess.run(f'python yolov5/detect.py --weights {latest_run}/weights/best.pt --img 640 --source drive/MyDrive/computervision/test/ --data yolov5/data/fox.yaml', shell=True, capture_output=True, text=True)
if latest_run:
    # COMANDO
    result = subprocess.run(f'python yolov5/detect.py --weights {latest_run}/weights/best.pt --img 640 --source drive/MyDrive/computervision/test/ --data yolov5/data/fox.yaml', shell=True, capture_output=True, text=True)
    print(result.stdout)
    print(result.stderr)
else:
    print("Não foi possível encontrar a pasta de treinamento mais recente.")
```

###Abaixo algumas conclusões:

- 🦊 O modelo conseguiu identificar raposas em diferentes posições e contextos, demonstrando sua robustez.
- 🌽 Também detectou espigas de milho **sem a palha**, ignorando grande parte do restante da planta.

💡 Um insight interessante foi observar que o modelo só detecta a parte descoberta do milho (a parte amarela), o que pode ser uma funcionalidade útil para o cenário real:

> **"Se o sistema estiver monitorando constantemente o milharal e identificar uma espiga fora da planta e sem palha no chão, isso pode ser interpretado como um indício de ataque de uma raposa, mesmo que o animal ainda não tenha sido detectado diretamente pela câmera."**

### Testando o modelo: Nesta etapa, mais uma vez estamos usando imagens nuca vistas para avaliarmos seu desempenho.

```python
def get_latest_train_run_folder():
    subfolders = [f.path for f in os.scandir('yolov5/runs/train') if f.is_dir()]
    latest_folder = max(subfolders, key=os.path.getctime, default=None)
    return latest_folder

latest_run = get_latest_train_run_folder()
result = subprocess.run(f'python yolov5/detect.py --weights {latest_run}/weights/best.pt --img 640 --source drive/MyDrive/computervision/test/ --data yolov5/data/fox.yaml', shell=True, capture_output=True, text=True)
if latest_run:
    # COMANDO
    result = subprocess.run(f'python yolov5/detect.py --weights {latest_run}/weights/best.pt --img 640 --source drive/MyDrive/computervision/test/ --data yolov5/data/fox.yaml', shell=True, capture_output=True, text=True)
    print(result.stdout)
    print(result.stderr)
else:
    print("Não foi possível encontrar a pasta de treinamento mais recente.")
```
